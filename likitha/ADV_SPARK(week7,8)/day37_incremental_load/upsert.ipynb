```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.master("local[*]").getOrCreate()
```

```python
target = spark.createDataFrame([
    (4, "Charmander", 0.7),
    (8, "Wartortle", 1.0)
], ["pokemon_id", "name", "height_m"])
```

```python
target.show()
```

```python
batch = spark.createDataFrame([
    (4, "Charmander", 0.8),
    (10, "Caterpie", 0.3)
], ["pokemon_id", "name", "height_m"])
```

```python
batch.show()
```

```python
target.unionByName(batch).show()
```

```python
result = (
    target.join(
        batch.select("pokemon_id"),
        on="pokemon_id",
        how="left_anti"
    )
    .unionByName(batch)
)
```

```python
result.orderBy("pokemon_id").show()
```

```python
spark.stop()
```
