```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.getOrCreate()
```

```python
sc=spark.sparkContext
```

```python
rdd1=sc.parallelize([1 ,2 ,3 ,4 ,5])
rdd1
```

```python
rdd1.getNumPartitions()
```

```python
rdd1.collect()
```

```python
rdd2=rdd1.map(lambda x : x*2)
rdd2.collect()
```

```python
rdd2=rdd1.filter(lambda x : x%2==0)
rdd2.collect()
```

```python
rdd=sc.parallelize(["hello world" , "hello spark" , "hello pyspark"])
```

```python
rdd1=rdd.flatMap(lambda x : x.split())
rdd1.collect()
```

```python
rdd2=rdd.map(lambda x : x.split())
rdd2.collect()
```

```python
rdd3=rdd1.map(lambda x : (x,1))
rdd3.collect()
```

```python
rdd3.reduceByKey(lambda a , b : a+b).collect()
```

```python
rdd4=rdd3.groupByKey()
rdd4.collect()
```

```python
list_of_sorted_group=list((key , list(values)) for key  , values in rdd4.collect())
print(list_of_sorted_group)
```

```python
rdd4.mapValues(lambda x: sum(x)).collect()
```
