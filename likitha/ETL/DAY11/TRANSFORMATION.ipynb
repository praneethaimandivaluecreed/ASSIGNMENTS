```python
import pandas as pd
```

```python
def extract_data():
    data = {
        "product_id": [101, 102, 103, 104, 105],
        "product_name": ["Laptop", "Mouse", "Keyboard", "Chair", "Monitor"],
        "category": [
            "Electronics",
            "Electronics",
            "Electronics",
            "Furniture",
            "Electronics"
        ],
        "price": [55000, 800, 1500, 7000, 12000],
        "quantity": [10, 25, 15, 5, 8],
        "status": ["active", "active", "inactive", "active", "active"]
    }

    df = pd.DataFrame(data)

    return df
```

```python
def clean_data(df):
    # Remove duplicate rows
    df = df.drop_duplicates()

    # Remove rows where important values are missing
    df = df.dropna(subset=["product_id", "product_name"])

    return df
```

```python
def filter_active_products(df):
    # Keep only active products
    df = df[df["status"] == "active"]

    return df
```

```python
def calculate_total_value(df):
    # Calculate total inventory value
    df["total_value"] = df["price"] * df["quantity"]

    return df
```

```python
def standardize_category(df):
    # Make category names consistent
    df["category"] = df["category"].str.strip().str.title()

    return df
```

```python
def select_required_columns(df):
    # Keep only columns required for final output
    columns = [
        "product_id",
        "product_name",
        "category",
        "price",
        "quantity",
        "total_value"
    ]

    return df[columns]
```

```python
def load_data(df):
    print("\nFinal transformed data:")
    print(df)
```

```python
def main():

    # 1. Extract
    df = extract_data()

    print("RAW DATA:")
    print(df)

    # 2. Transform
    df = clean_data(df)
    df = filter_active_products(df)
    df = calculate_total_value(df)
    df = standardize_category(df)
    df = select_required_columns(df)

    # 3. Load
    load_data(df)
```

```python
main()
```
