```python
orders_df = orders_df.withColumn(
    "order_date",
    sf.to_date("order_date")
)
```

```python
orders_df.show()
orders_df.printSchema()
```

Assign a sequential order number to every order for each customer based on the order date.

Expected columns:
customer_id
order_id
order_date
amount
order_number

```python
window=Window.partitionBy("customer_id").orderBy(sf.col("order_date"))

df=orders_df.withColumn("rn" , sf.row_number().over(window))
df.select("customer_id" , "order_id" , "order_date" , "amount" , "rn").show()
```

For every customer, find the amount spent on their previous order and calculate how much their current order amount differs from the previous order.

```python
window=Window.partitionBy("customer_id").orderBy(sf.col("order_date"))
df=orders_df.withColumn("previous_amount" ,sf.lag("amount").over(window))
df=df.withColumn("amount_diff" , sf.col("amount")-sf.col("previous_amount"))
df.select("order_id" , "amount" , "previous_amount"  , "amount_diff").show()
```

Calculate the cumulative amount spent by each customer after every order, based on the order date.

```python
window=Window.partitionBy("customer_id").orderBy(sf.col("order_date")).rowsBetween(Window.unboundedPreceding , Window.currentRow)

df=orders_df.withColumn("cumulative_price" , sf.sum("amount").over(window))
df.select("order_id" , "amount" , "cumulative_price").show()
```

Find the top 2 highest-value orders within each city.

```python
window=Window.partitionBy("city").orderBy(sf.col("amount").desc())
df=orders_df.withColumn("rn" , sf.rank().over(window))
df.filter(sf.col("rn")<=2).select("city" , "order_id" , "customer_name", "product" , "amount" ,"rn").show()
```
