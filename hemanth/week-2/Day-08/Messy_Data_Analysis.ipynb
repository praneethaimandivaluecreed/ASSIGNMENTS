```python
import pandas as pd

df = pd.read_csv('consumer_shopping_behavior_survey.csv')
#understanding the given dataset
print(df.head()) #observing first five rows to get a galance of given dataset
print(df.tail())
print(df.columns)
print(df.dtypes)
print(df.info())
print(df.describe())
```

```python
#checking missing values
missing_values = pd.DataFrame({
    "Missing_values": df.isnull().sum(),
    "Missing_values_percent":df.isnull().mean() * 100
})
print(missing_values)
```

```python
#checking duplicate values
print(df.duplicated().sum())

#finding the unique values in each column, so that it helps what to replace null values
#in categorical columns

for column in df.columns:
  print(f'\n {column} ')
  print(df[column].unique())
```

#Performing the cleaning,transforming and analysis on age column

```python
#Transforming the data
df["Age"] = pd.to_numeric(df["Age"],errors='coerce')
print(df["Age"].value_counts(dropna=False))
```

```python
print(df["Age"].describe())
```

```python
#finding median value and fill the null values with median value and finding null value
median_value = df["Age"].agg("median")
df["Age"] = df['Age'].fillna(median_value)
print(df["Age"].isnull().sum())
```

```python
#Gender column
print(df["Gender"].value_counts(dropna=False))
```

```python
#replacing null values with "Unkown"
df["Gender"] = df["Gender"].fillna("Unkow")
print(df['Gender'].value_counts())
print(df["Gender"].isnull().sum())
```

```python
#TimeStamp column
print(df['Timestamp'].isnull().sum())
print(df["Timestamp"].head())#to find the time format of values
print(df["Timestamp"].tail())
df["Timestamp"] = pd.to_datetime(df["Timestamp"],format = '%d/%m/%Y %H:%M:%S')
#formatting all the timestamps with correct format
```

```python
print(df.select_dtypes(include='number').columns)
print(df.select_dtypes(include='object').columns)
```

```python
num_cols = df.select_dtypes(include='number').columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
print(df[num_cols].isnull().sum())
#finding the numerical values and replacing the each numerical column null values with its column median value
#printing all numerical value columns null count
```

```python
cat_cols = df.select_dtypes(include='object').columns

df[cat_cols] = df[cat_cols].fillna("Unknown")
print(df[cat_cols].isnull().sum())

#finding the categorical values and replacing the each categorical column null values with its column median value
#printing all categorcial value columns null count
```

```python
#final checking of null values
print(df.isnull().sum().sum())
```
