```python
import pandas as pd
data = {
    "Name": ["Rahul", "Priya", "Arjun"],
    "Age": [25, None, 30]
}
df = pd.DataFrame(data)
print(df)# None is converted into NaN

data = [25, None, 30]
print(type(data[1]))

print(df["Age"].dtype)
#so pandas converted the None into NaN while interpeting data type of Age coloumn
#Here it is converted into float because Traditional NumPy integer dtype doesn't have a built-in NaN


#understanding Pandas nullable integer

df = pd.DataFrame({
    "Age": [25, None, 30]
})


df['Age'] = df["Age"].astype("Int64")

print(df)
print(df["Age"].dtype)


```

```python

data = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score': [np.nan, 40, 80, 98]}
df = pd.DataFrame(data)

print(df.isnull())
print("\nchecking with .isna():\n")
print(df.isna())

print("\n checking with .notnull()\n")
print(df.notnull())
```

```python
df = pd.read_csv("employees.csv")

nullvalues = pd.isnull(df['Gender'])
print(nullvalues)

print()
notnullvalues = pd.notnull(df['Gender'])
print(notnullvalues)
```

```python
#Filling missing values
data = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score': [np.nan, 40, 80, 98]}
df = pd.DataFrame(data)
print(df)
df = df.fillna(0)
print("\nDataFrame after filling values\n")
print(df)
```

```python
#replace function
df = pd.DataFrame({
    "Name": ["Ravi", "Hemanth", "Priya"],
    "Department": ["IT", "HR", "IT"]
})
print("\n Before replacing \n")
print(df)
df = df.replace("IT","Information Technology")
print("\n After replacing\n")
print(df)
```

```python
#interpolate function
df = pd.DataFrame({
    "Age": [20, 22, None, 26, 28]
})

print(df)
print(df.interpolate())

#Pandas estimates the missing value halfway between.
```

```python
#dropna function
data = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, np.nan, 45, 56],
        'Third Score': [52, 40, 80, 98],
        'Fourth Score': [np.nan, np.nan, np.nan, 45]}
df = pd.DataFrame(data)
print(df)
print('\n after using dropna func \n')
print(df.dropna())

print("\n with how ='all \n")
print(df.dropna(how='all'))

print("\ndropping rows with atleast one null value\n")
print(df.dropna(axis = 1))
```

```python
#drop_duplicates function

data = {
    "Name": ["Alice", "Bob", "Alice", "David"],
    "Age": [25, 30, 25, 40],
    "City": ["NY", "LA", "NY", "Chicago"]
}
df = pd.DataFrame(data)
print(df)
print()
print(df.drop_duplicates())
```

```python
#astype method

df = pd.DataFrame({'A': ['1', 'two'], 'B': ['3.0', '4.5']})

print(df)
print(df.dtypes)
print()
df["A"] = df["A"].astype(int, errors="ignore")
#method tries to convert column 'A' to integer, but the value 'two' is not numeric.
#Since errors='ignore' is used, pandas skips the conversion
# and retains the original data types.
print(df)
print(df.dtypes)

df["A"] = df["A"].astype(int,errors = "raise")
#Pandas will raise an error when it encounters "two".
print(df)
print(df.dtypes)
```

```python
#to_datatime function
import pandas as pd
data = {
    "Date": [
        "25/08/2025",
        "31/12/2025",
        "15/01/2026",
        "invalid",
        "2026-03-20"
    ],
    "Value": [100, 200, 300, 400, 500]
}
df = pd.DataFrame(data)
res1 = pd.to_datetime(df['Date'],errors = 'coerce')
res2 = pd.to_datetime(df['Date'],errors = 'ignore')
print('\n coerce:\n',res1)
print('\n ignore:\n',res2)

try:
    res3 = pd.to_datetime(
        df["Date"],
        errors="raise",
        format="mixed",
        dayfirst=True
    )

    print("\nRaise:\n", res3)

except ValueError as e:
    print("\nRaise produced an error:")
    print(e)



```
